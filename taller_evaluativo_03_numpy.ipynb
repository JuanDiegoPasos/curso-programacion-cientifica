**Ejercicio 1**

Se crea la tabla de lecturas de los cinco sensores y se describe con sus atributos.

```python
import numpy as np

# datos crudos
lecturas_crudas = np.array([
    [10.0, 14.5, 22.0, 30.5, 28.0, 18.0, 9.5],
    [12.0, 20.0, 35.0, 48.0, 41.0, 25.0, 14.0],
    [8.0, 11.0, 19.0, 27.0, 33.0, 21.0, 6.0],
    [15.0, 24.0, 31.0, 44.0, 52.0, 36.0, 17.0],
    [5.0, 9.0, 16.0, 22.0, 45.0, 30.0, 4.0],
])

# atributos
forma = lecturas_crudas.shape
dimensiones = lecturas_crudas.ndim
total = lecturas_crudas.size
tipo = lecturas_crudas.dtype

# días
dias = np.arange(1, 8)

# calibración
pendientes = np.array([1.2, 0.9, 1.1, 1.0, 1.3])
interceptos = np.array([-0.5, 1.0, -2.0, 0.5, -1.5])

print("Forma:", forma)
print("Dimensiones:", dimensiones)
print("Total de elementos:", total)
print("Tipo de dato:", tipo)
print("Días:", dias)
print("Pendientes:", pendientes)
print("Interceptos:", interceptos)
```

**Ejercicio 2**


Se leen celdas, columnas, filas y un bloque de la tabla con indexación y slicing.

```python
# una celda
lectura_c_dia4 = lecturas_crudas[2, 3]

# índice negativo
ultimo_primer = lecturas_crudas[-1, 0]

# columna completa
dia5 = lecturas_crudas[:, 4]

# fila completa
semana_b = lecturas_crudas[1, :]

# bloque 3x4
bloque = lecturas_crudas[1:4, 2:6]

print("SBC-C en el día 4:", lectura_c_dia4)
print("Último sensor en el día 1:", ultimo_primer)
print("Día 5:", dia5)
print("Semana de SBC-B:", semana_b)
print("Bloque:")
print(bloque)
print("Forma del bloque:", bloque.shape)
```

**Ejercicio 3 **

Se descartar el centinela -999.0 y el ruido de la campaña de SBC-A usando máscaras.

```python
campana_sbc_a = np.array([10.0, -999.0, 25.0, 40.0, -999.0, 8.0, 52.0, 600.0])

# centinelas
faltantes = campana_sbc_a == -999.0

# rango válido
es_util = (campana_sbc_a >= 0) & (campana_sbc_a <= 500.0)

# filtrar
lecturas_utiles = campana_sbc_a[es_util]
promedio_utiles = lecturas_utiles.mean()

print("Lecturas faltantes:", faltantes.sum())
print("Lecturas útiles:", lecturas_utiles)
print("Promedio de lecturas útiles:", promedio_utiles)
```

**Ejercicio 4**

lecturas_crudas tiene forma (5, 7) y pendientes tiene forma (5,). NumPy compara las formas de derecha a izquierda, o sea 7 con 5, y como no son iguales da error. Al convertir pendientes`en una columna de forma (5, 1), NumPy repite la pendiente de cada sensor en los 7 días y ya se puede multiplicar.

```python
# columnas (5, 1)
pendientes_col = pendientes.reshape(5, 1)
interceptos_col = interceptos[:, np.newaxis]

# calibrar la red
calibradas = lecturas_crudas * pendientes_col + interceptos_col

# calibrar campaña
campana_calibrada = lecturas_utiles * 1.2 + (-0.5)
promedio_calibrado_sbc_a = campana_calibrada.mean()

print("Forma de calibradas:", calibradas.shape)
print("Fila de SBC-A calibrada:", calibradas[0])
print("Campaña calibrada de SBC-A:", campana_calibrada)
print("Promedio calibrado de SBC-A:", promedio_calibrado_sbc_a)
```

**Ejercicio 5**

Se resume la red completa, por sensor (axis=1) y por día (axis=0).

```python
# red completa
promedio_general = calibradas.mean()

# por sensor
promedio_por_sensor = calibradas.mean(axis=1)

# por día
promedio_por_dia = calibradas.mean(axis=0)

# máximos
maximo_por_sensor = calibradas.max(axis=1)

# sobre el límite
media_sobre_limite = calibradas[calibradas > 35.4].mean()

print("Promedio general:", np.round(promedio_general, 2))
print("Promedio por sensor:", np.round(promedio_por_sensor, 2))
print("Promedio por día:", np.round(promedio_por_dia, 2))
print("Máximo por sensor:", np.round(maximo_por_sensor, 2))
print("Media sobre el límite:", np.round(media_sobre_limite, 2))
```

**Ejercicio 6**
Se mide cuánto se pasó cada sensor del límite de 35.4 µg/m³ y se marcan las lecturas en alerta.

```python
# cuánto se pasó
excedente = np.maximum(calibradas - 35.4, 0)
excedente_por_sensor = np.round(excedente.sum(axis=1), 1)

# distancia al límite
desvio_sensor_b = np.round(np.abs(calibradas[1] - 35.4), 1)

# copia con reemplazo
en_alerta = calibradas.copy()
en_alerta[en_alerta <= 35.4] = 0.0

print("Excedente por sensor:", excedente_por_sensor)
print("Desvío de SBC-B:", desvio_sensor_b)
print("Celdas en alerta:", (en_alerta > 0).sum())
```

**Ejercicio 7**

Se cuentan las lecturas de cada categoría de la escala oficial y se identifica el día más contaminado.

```python
# categorías
buenas = (calibradas <= 12.0).sum()
moderadas = ((calibradas > 12.0) & (calibradas <= 35.4)).sum()
daninas_sensibles = ((calibradas > 35.4) & (calibradas <= 55.4)).sum()
daninas_salud = (calibradas > 55.4).sum()

# día pico
dia_pico = dias[promedio_por_dia == promedio_por_dia.max()][0]

print(f"Lecturas por categoría: Buena={buenas}, Moderada={moderadas}, Dañina para grupos sensibles={daninas_sensibles}, Dañina para la salud={daninas_salud}")
print(f"Día más contaminado: {dia_pico} con {np.round(promedio_por_dia.max(), 1)} ug/m3")
```

## Reto de extensión
Se obtienen los mismos cuatro conteos con `np.digitize` y se comparan con los del Ejercicio 7.

```python
# índices
indices = np.digitize(calibradas, [12.0, 35.4, 55.4], right=True)

buenas_digitize = (indices == 0).sum()
moderadas_digitize = (indices == 1).sum()
daninas_sensibles_digitize = (indices == 2).sum()
daninas_salud_digitize = (indices == 3).sum()

print("Conteos con digitize:", buenas_digitize, moderadas_digitize, daninas_sensibles_digitize, daninas_salud_digitize)
print("Coinciden con el Ejercicio 7:", (buenas_digitize, moderadas_digitize, daninas_sensibles_digitize, daninas_salud_digitize) == (buenas, moderadas, daninas_sensibles, daninas_salud))
```
