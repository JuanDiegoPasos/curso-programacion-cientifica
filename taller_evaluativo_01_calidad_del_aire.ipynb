**Ejercicio 1**

Declarar metadatos de la estación con su tipo de dato y se verificar cada tipo.

```python
# texto
codigo_estacion = "EST-01"
sector = "Norte"

# enteros
altitud_m = 1495
numero_lecturas = 305
factor_conversion = 1000

# decimal
latitud = 6.28

# booleano
sensor_calibrado = True

print("codigo_estacion:", type(codigo_estacion))
print("sector:", type(sector))
print("altitud_m:", type(altitud_m))
print("numero_lecturas:", type(numero_lecturas))
print("latitud:", type(latitud))
print("sensor_calibrado:", type(sensor_calibrado))
print("factor_conversion:", type(factor_conversion))
```

**Ejercicio 2**

Convertir la lectura del sensor a ug/m3 y calculanrlos errores frente al valor del patrón.

```python
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

# conversión
lectura_sensor_ug = lectura_sensor_mg * factor_conversion

# errores
error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = error_absoluto / valor_patron_ug * 100
error_cuadratico = error_absoluto ** 2

print("Lectura del sensor en ug/m3:", lectura_sensor_ug)
print("Error absoluto:", round(error_absoluto, 2))
print("Error relativo porcentual:", round(error_relativo_porcentual, 2))
print("Error cuadrático:", round(error_cuadratico, 2))
```

**Ejercicio 3**

Calcula cuánto tiempo cubren las lecturas descargadas.

```python
lecturas_por_hora = 12

# horas
horas_completas = numero_lecturas // lecturas_por_hora
lecturas_sobrantes = numero_lecturas % lecturas_por_hora

# días
dias_completos = horas_completas // 24
horas_sobrantes = horas_completas % 24

print("Horas completas:", horas_completas)
print("Lecturas sobrantes:", lecturas_sobrantes)
print("Días completos:", dias_completos)
print("Horas sobrantes:", horas_sobrantes)
```

**Ejercicio 4**

Se evalúar si la lectura es válida sin usar if.

```python
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

# condiciones
es_dato_faltante = lectura == codigo_dato_faltante
esta_en_rango = lectura >= 0 and lectura <= limite_superior_sensor
lectura_valida = esta_en_rango and not es_dato_faltante

print("Es dato faltante:", es_dato_faltante)
print("Está en rango:", esta_en_rango)
print("Lectura válida:", lectura_valida)
```

**Ejercicio 5**

Clasificar la lectura con una cadena según la escala oficial usando if, elif, else.

```python
# escala oficial
if lectura <= 12.0:
    categoria = "Buena"
elif lectura <= 35.4:
    categoria = "Moderada"
elif lectura <= 55.4:
    categoria = "Dañina para grupos sensibles"
else:
    categoria = "Dañina para la salud"

print("Categoría:", categoria)
```

**Ejercicio 6**

recorrer datos descartando los centinelas con un continue y despues calcular el promedio de las lecturas válidas.

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]

# acumuladores
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for medicion in lecturas_pm25:
    # saltar centinela
    if medicion == codigo_dato_faltante:
        lecturas_descartadas += 1
        continue
    lecturas_validas.append(medicion)
    suma_pm25 += medicion

# promedio
promedio_pm25 = suma_pm25 / len(lecturas_validas)

print("Lecturas válidas:", len(lecturas_validas))
print("Lecturas descartadas:", lecturas_descartadas)
print("Promedio:", promedio_pm25)
```

**Ejercicio 7**

Calcular extremos, rango y desviación estándar de las lecturas válidas.

```python
# extremos
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for medicion in lecturas_validas:
    if medicion > maximo_pm25:
        maximo_pm25 = medicion
    if medicion < minimo_pm25:
        minimo_pm25 = medicion

rango_pm25 = maximo_pm25 - minimo_pm25

# dispersión
suma_desviaciones = 0.0

for medicion in lecturas_validas:
    suma_desviaciones += (medicion - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print("Máximo:", maximo_pm25)
print("Mínimo:", minimo_pm25)
print("Rango:", rango_pm25)
print("Desviación estándar:", round(desviacion_estandar, 2))
```

**Ejercicio 8**

Contar las horas que tarda la concentración en volver a la categoría Buena.

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0

# contador
horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion = concentracion * factor_decaimiento
    horas_transcurridas += 1

print("Horas transcurridas:", horas_transcurridas)
print("Concentración final:", round(concentracion, 2))
```

**Ejercicio 9**

Recorrer los diccionarios de las estaciones para resumir cada una con su promedio, categoría y humedad.

```python
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte", "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur", "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for estacion in registros_estaciones:
    # desempaquetar
    latitud, longitud = estacion["coordenadas"]

    # promedio
    suma_estacion = 0.0
    for valor in estacion["lecturas"]:
        suma_estacion += valor
    promedio_estacion = suma_estacion / len(estacion["lecturas"])

    # humedad
    humedad = estacion.get("humedad_relativa", "no disponible")

    # categoría
    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(estacion["codigo"], "|", estacion["sector"], "| lat", latitud, "| lon", longitud, "| promedio", round(promedio_estacion, 2), "|", categoria_estacion, "| humedad", humedad)
```
